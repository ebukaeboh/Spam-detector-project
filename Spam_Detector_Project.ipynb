```python

# Import the tools we need
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# Download the spam messages data
url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"
data = pd.read_csv(url, sep='\t', header=None, names=['label', 'message'])

# Show the first 5 rows
print(data.head())
print(data['label'].value_counts())
```

```python
# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(data['message'], data['label'], test_size=0.2, random_state=42)

# Turn words into numbers the computer can understand
vectorizer = TfidfVectorizer()
X_train_numbers = vectorizer.fit_transform(X_train)
X_test_numbers = vectorizer.transform(X_test)

# Create and train the model
model = MultinomialNB()
model.fit(X_train_numbers, y_train)

# Check how good it is
prediction = model.predict(X_test_numbers)
print("Accuracy:", accuracy_score(y_test, prediction))
```
